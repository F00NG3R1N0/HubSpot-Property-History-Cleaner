```python
# Identify the first two columns as identifiers
identifier_columns = ['Record ID', 'Email']

# Initialize an empty list to store the stacked DataFrames
stacked_df_list = []

# Iterate through the columns in pairs, starting from the third column
num_columns = len(df.columns)
for i in range(2, num_columns, 4):
    # Select the current pair of columns along with the identifier columns
    pair_df = df[identifier_columns + [df.columns[i], df.columns[i + 1], df.columns[i + 2], df.columns[i + 3]]]

    # Drop rows where the second column in the pair has NaN
    pair_df = pair_df.dropna(subset=[df.columns[i + 1]])

    # Rename the columns to have consistent names
    pair_df.columns = identifier_columns + ['Value', 'Change Date (US/Eastern)', 'Change Source', 'Changed By']
    
    # Append the pair_df to the list
    stacked_df_list.append(pair_df)

# Concatenate all the DataFrames in the list
final_df = pd.concat(stacked_df_list, ignore_index=True)

# Output the result
final_df
```

```python
# Identify the first two columns as identifiers
identifier_columns = ['Record ID', 'Email']

# Initialize an empty list to store the stacked DataFrames
stacked_df_list = []

# Iterate through the columns in pairs, starting from the third column
num_columns = len(df.columns)
for i in range(2, num_columns, 4):
    # Check if there are enough columns left to form a complete quartet
    if i + 3 >= num_columns:
        break  # Exit the loop if there are not enough columns left

    # Select the current quartet of columns along with the identifier columns
    quartet_df = df[identifier_columns + [df.columns[i], df.columns[i + 1], df.columns[i + 2], df.columns[i + 3]]]
    
    # Drop rows where the fourth column in the quartet has NaT
    quartet_df = quartet_df.dropna(subset=[df.columns[i + 3]])

    # Rename the columns to have consistent names
    quartet_df.columns = identifier_columns + ['Campaign UTM Current Value', 'Campaign UTM Change Date (US/Eastern)', 'Campaign UTM Change Source', 'Campaign UTM Changed By']
    
    # Append the pair_df to the list
    stacked_df_list.append(quartet_df)

# Concatenate all the DataFrames in the list
final_df = pd.concat(stacked_df_list, ignore_index=True)

# Output the result
final_df
```
